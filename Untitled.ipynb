```python
# QUESTION A SOLUTION 
# initially what we do is we  Load and prepare the data
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Loading the dataset
df = pd.read_csv('C:/Users/DELL/Kubra/Downloads/Carseats-4.csv') 

# Encode categorical variables
df_encoded = df.copy()
label_encoders = {}
for col in ['ShelveLoc', 'Urban', 'US']:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col])
    label_encoders[col] = le

# Defining the  features and target
X = df_encoded.drop('Sales', axis=1)
y = df_encoded['Sales']

# Splitting the data into train and test sets (70/30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# QUESTION B SOLUTION and the results are interpreted 
#  regression tree and evaluation 
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

# Fitting the  regression tree
tree_model = DecisionTreeRegressor(random_state=42)
tree_model.fit(X_train, y_train)


plt.figure(figsize=(20, 10))
plot_tree(tree_model, feature_names=X.columns, filled=True, rounded=True, max_depth=3)
plt.title("Regression Tree (First 3 Levels)")
plt.show()

# Predicting  and calculating  test MSE
y_pred_tree = tree_model.predict(X_test)
test_mse_tree = mean_squared_error(y_test, y_pred_tree)
print(f"Test MSE (Regression Tree): {test_mse_tree:.4f}")
```

```python
# QUESTION C SOLUTION 
# using  cross-validation to prune the trees
from sklearn.model_selection import cross_val_score
import numpy as np

depths = range(1, 21)
cv_scores = []

for d in depths:
    model = DecisionTreeRegressor(max_depth=d, random_state=42)
    mse = -cross_val_score(model, X_train, y_train, scoring='neg_mean_squared_error', cv=10).mean()
    cv_scores.append(mse)


best_depth = depths[np.argmin(cv_scores)]
print(f"Optimal tree depth from cross-validation: {best_depth}")

# Trained  a pruned tree
pruned_tree = DecisionTreeRegressor(max_depth=best_depth, random_state=42)
pruned_tree.fit(X_train, y_train)
y_pred_pruned = pruned_tree.predict(X_test)
test_mse_pruned = mean_squared_error(y_test, y_pred_pruned)
print(f"Test MSE (Pruned Tree): {test_mse_pruned:.4f}")
print("yes, pruning the tree improved the test mse")
```

```python
# QUESTION D SOLUTION 
# Bagging
from sklearn.ensemble import BaggingRegressor

bagging_model = BaggingRegressor(random_state=42, n_estimators=100)
bagging_model.fit(X_train, y_train)

y_pred_bagging = bagging_model.predict(X_test)
test_mse_bagging = mean_squared_error(y_test, y_pred_bagging)
print(f"Test MSE (Bagging): {test_mse_bagging:.4f}")
```

```python
# Importing the  libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

# Loading the  dataset
df = pd.read_csv("C:/Users/DELL/Kubra/Downloads/Carseats-4.csv")  

# Encoding the categorical features
df_encoded = df.copy()
label_encoders = {}
for col in ['ShelveLoc', 'Urban', 'US']:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col])
    label_encoders[col] = le

# Features and target
X = df_encoded.drop("Sales", axis=1)
y = df_encoded["Sales"]

# Train or the test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Test MSE for different values of m
m_values = list(range(1, X.shape[1] + 1))
mse_list = []

for m in m_values:
    rf = RandomForestRegressor(n_estimators=100, max_features=m, random_state=42)
    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_list.append(mse)

# Plotting MSE vs m
plt.figure(figsize=(10, 6))
plt.plot(m_values, mse_list, marker='o', linestyle='-')
plt.xlabel("Number of Features Considered at Each Split (m)")
plt.ylabel("Test MSE")
plt.title("Effect of 'm' on Random Forest Test MSE")
plt.xticks(m_values)
plt.grid(True)
plt.show()
```

```python
# Q2.WE are going to analyse the above data using boosting and the result will be interpreted.
# Importing the  Gradient Boosting
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
import pandas as pd

# Training  the Gradient Boosting model
boost_model = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)
boost_model.fit(X_train, y_train)

# Predicting  and evaluating the  test MSE
y_pred_boost = boost_model.predict(X_test)
test_mse_boost = mean_squared_error(y_test, y_pred_boost)

# Feature importances
boost_importances = pd.Series(
    boost_model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print("Test MSE (Boosting):", test_mse_boost)
print("\nFeature Importances:")
print(boost_importances)
```

To further improve prediction performance, we applied the Gradient Boosting technique to the Carseats dataset. 
With 100 estimators, learning rate 0.1, and depth 3, the boosting model achieved a test MSE of 1.928, which is superior to all other models like bagging and random forests. 
This means that boosting can better generalize sophisticated interactions and reduce bias more. Price, ShelveLoc, Age, CompPrice, and Advertising were the strongest model features, with Price being the best-performing single predictor. 
These results do go some way to establishing the strength of the boosting methodology to regression problems when appropriately tuned to the correct hyperparameters.
